```python
# Prepping the environment and performing EDA on a CSV dataset
# Imports
import sys, os, csv
print("Python executable:", sys.executable)
print("Python version:", sys.version.splitlines()[0])
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
from pathlib import Path

# Check dataset path
DATA_PATH = os.path.join("data", "full_dataset-release.csv")
print("Expected CSV path:", DATA_PATH)
print("File exists:", os.path.exists(DATA_PATH))
if os.path.exists(DATA_PATH):
    print("Size (bytes):", os.path.getsize(DATA_PATH))
else:
    print("File is missing at location ", DATA_PATH)

# Display settings
print("Setting up display options...")
pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.4f}".format)
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
print("Display options set.")

# Load dataset
print("Loading dataset...")
DATA_PATH = Path("data") / "full_dataset-release.csv"
df = pd.read_csv(
    DATA_PATH,
    quoting=csv.QUOTE_ALL,
    encoding='utf-8',
    on_bad_lines='skip',
    engine='python'
)
print("Loaded:", DATA_PATH, "shape:", df.shape)
print("Dataset loaded.")

# Quick peek
display(df.head())
display(df.tail())

# Data cleaning
print("Cleaning data...")
# Clean column names
df.columns = df.columns.str.strip().str.replace(r"[\r\n]+", "", regex=True)
# Clean all string columns (especially TWEET)
df = df.apply(lambda col: col.str.replace(r"[\r\n]+", " ", regex=True).str.strip() if col.dtype == "object" else col)
# Convert DATE column
df["DATE"] = pd.to_datetime(df["DATE"], format="%d/%m/%Y", errors="coerce")
# Drop duplicates and rows missing key info
df.drop_duplicates(inplace=True)
df.dropna(subset=["DATE", "STOCK", "LAST_PRICE"], inplace=True)
# Convert numeric columns
numeric_cols = df.columns.drop(["TWEET", "STOCK", "DATE"])
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors="coerce")
# Add sentiment label
df["LSTM_SENTIMENT"] = df["LSTM_POLARITY"].map({-1: "Negative", 0: "Neutral", 1: "Positive"})
# Save clean CSV
df.to_csv("data/cleaned_data.csv", index=False, lineterminator="\n")
print("Data cleaned.")

print("Null Tweets - \n", df[df["TWEET"].isnull()])
print("DataFrame shape:", df.shape)
print("Columns ", df.columns.tolist())
print("Info", df.info())
print("Describe", df.describe(include="all").T)
print("Head ", df.head())
print("Tail ", df.tail())

# Info & summary
print("\nDataFrame.info():")
display(df.info(verbose=True))

print("\nDescribe (numeric):")
display(df.describe().T)

print("\nDescribe (all):")
display(df.describe(include="all").T)

# Missing values
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(3)
missing_tbl = pd.concat([missing, missing_pct], axis=1)
missing_tbl.columns = ["missing_count", "missing_pct"]
display(missing_tbl[missing_tbl["missing_count"] > 0])

# Unique values and potential identifiers
unique_counts = df.nunique(dropna=False).sort_values()
display(unique_counts.to_frame(name="unique_count").tail(30))

# Duplicates
dups = df.duplicated().sum()
print(f"\nDuplicate rows: {dups}")
if dups:
    display(df[df.duplicated(keep=False)].head())

# Data types
dtypes = df.dtypes.value_counts()
print("\nData types distribution:")
display(dtypes)

# Detect and parse probable datetime columns (names containing 'date' or 'time')
date_candidates = [c for c in df.columns if ("date" in c.lower() or "time" in c.lower())]
parsed_dates = {}
for c in date_candidates:
    try:
        parsed = pd.to_datetime(df[c], errors="coerce", infer_datetime_format=True)
        non_null = parsed.notna().sum()
        if non_null:
            parsed_dates[c] = {"parsed_non_null": int(non_null), "min": parsed.min(), "max": parsed.max()}
            df[c + "_parsed"] = parsed
    except Exception:
        pass
if parsed_dates:
    print("\nParsed date/time candidates:")
    display(pd.DataFrame(parsed_dates).T)
else:
    print("\nNo parseable date/time columns detected by heuristic.")

# Numeric columns EDA: histograms and skew
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"\nNumeric columns ({len(num_cols)}): {num_cols[:20]}")
if num_cols:
    sample_nums = num_cols if len(num_cols) <= 12 else num_cols[:12]
    df[sample_nums].hist(bins=30, layout=(int(np.ceil(len(sample_nums)/3)), 3), figsize=(12, 8))
    plt.tight_layout()
    plt.show()

    skewness = df[num_cols].skew().sort_values(ascending=False)
    display(skewness.to_frame(name="skew").head(20))

    # Correlation heatmap (limit size)
    corr = df[num_cols].corr()
    if corr.shape[0] <= 50:
        mask = np.triu(np.ones_like(corr, dtype=bool))
        plt.figure(figsize=(min(14, corr.shape[0]), min(12, corr.shape[0])))
        sns.heatmap(corr, mask=mask, cmap="coolwarm", center=0, annot=False, fmt=".2f")
        plt.title("Numeric correlation matrix")
        plt.show()
    else:
        print("Too many numeric columns to plot full correlation matrix (>{} cols).".format(50))

# Categorical columns: top value counts
cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
print(f"\nCategorical columns ({len(cat_cols)}): {cat_cols[:20]}")
for c in cat_cols[:20]:
    vc = df[c].value_counts(dropna=False).head(10)
    print(f"\nTop values for {c}:")
    display(vc)

# Save small EDA artifacts
out_dir = Path("eda_outputs")
out_dir.mkdir(exist_ok=True)
missing_tbl.to_csv(out_dir / "missing_summary.csv")
unique_counts.to_frame("unique_count").to_csv(out_dir / "unique_counts.csv")
print(f"\nSaved missing_summary.csv and unique_counts.csv to {out_dir}/")

# Final sample
print("\nRandom sample:")
display(df.sample(min(10, len(df)), random_state=42))
```

```python
# Extracting the information from the data
# Identify top 10 most frequently mentioned companies
top_10_stocks = df["STOCK"].value_counts().head(10).index.tolist()

# Create output directory for charts
chart_dir = Path("eda_outputs") / "stock_sentiment_charts"
chart_dir.mkdir(parents=True, exist_ok=True)

# Plotting density function
def plot_stock_with_sentiment_density(df, stock_name):
    stock_df = df[df["STOCK"] == stock_name].copy()
    stock_df = stock_df.dropna(subset=["DATE", "LAST_PRICE", "LSTM_SENTIMENT"])
    stock_df.sort_values("DATE", inplace=True)
    stock_df = stock_df[stock_df["DATE"] >= pd.Timestamp("2018-01-01")]

    # Aggregate sentiment counts per day
    sentiment_counts = stock_df.groupby(["DATE", "LSTM_SENTIMENT"]).size().unstack(fill_value=0)

    fig, ax1 = plt.subplots(figsize=(12, 6))
    fig.savefig(chart_dir / f"{stock_name}_sentiment_density.png")

    # Plot stock price
    ax1.plot(stock_df["DATE"], stock_df["LAST_PRICE"], color="blue", label="Stock Price", linewidth=2)

    # Add 7-day moving average
    stock_df["MA_7"] = stock_df["LAST_PRICE"].rolling(window=7).mean()
    ax1.plot(stock_df["DATE"], stock_df["MA_7"], label="7-Day MA", color="orange", linestyle="--")

    # Format x-axis
    ax1.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax1.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    plt.xticks(rotation=45)

    ax1.set_ylabel("Last Price")
    ax1.set_xlabel("Date")
    ax1.set_title(f"{stock_name} Stock Performance with Sentiment Density")

    # Plot sentiment density as bars
    ax2 = ax1.twinx()
    ax2.bar(sentiment_counts.index, sentiment_counts.get("Positive", 0), color="green", alpha=0.3, label="Positive Tweets")
    ax2.bar(sentiment_counts.index, -sentiment_counts.get("Negative", 0), color="red", alpha=0.3, label="Negative Tweets")
    ax2.set_ylabel("Tweet Volume (±)")
    ax2.axhline(0, color="gray", linewidth=0.5)

    # Combine legends
    lines, labels = ax1.get_legend_handles_labels()
    bars, bar_labels = ax2.get_legend_handles_labels()
    ax1.legend(lines + bars, labels + bar_labels)

    plt.tight_layout()
    plt.show()

# Plotting line and scatter function
def plot_stock_sentiment(df, stock_name):
    stock_df = df[df["STOCK"] == stock_name].copy()
    stock_df.sort_values("DATE", inplace=True)
    stock_df = stock_df.dropna(subset=["DATE", "LAST_PRICE", "LSTM_SENTIMENT"])
    stock_df = stock_df[stock_df["DATE"] >= pd.Timestamp("2018-01-01")]

    # Aggregate sentiment per day
    sentiment_daily = stock_df.groupby(["DATE", "LSTM_SENTIMENT"]).size().unstack(fill_value=0)
    sentiment_daily["Dominant"] = sentiment_daily.apply(
        lambda row: "Positive" if row.get("Positive", 0) >= row.get("Negative", 0) else "Negative",
        axis=1
    )

    # Get average price per day
    price_daily = stock_df.groupby("DATE")["LAST_PRICE"].mean()

    # Merge sentiment and price
    plot_df = pd.DataFrame({
        "LAST_PRICE": price_daily,
        "Dominant_Sentiment": sentiment_daily["Dominant"]
    }).reset_index()

    fig, ax = plt.subplots(figsize=(12, 6))
    fig.savefig(chart_dir / f"{stock_name}_daily_sentiment.png")

    # Plot stock price line
    ax.plot(plot_df["DATE"], plot_df["LAST_PRICE"], label="Stock Price", color="blue", linewidth=2)

    # Add 7-day moving average
    plot_df["MA_7"] = plot_df["LAST_PRICE"].rolling(window=7).mean()
    ax.plot(plot_df["DATE"], plot_df["MA_7"], label="7-Day MA", color="orange", linestyle="--")

    # Format x-axis
    import matplotlib.dates as mdates
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    plt.xticks(rotation=45)

    # Plot one point per day based on dominant sentiment
    for sentiment, color in [("Positive", "green"), ("Negative", "red")]:
        subset = plot_df[plot_df["Dominant_Sentiment"] == sentiment]
        ax.scatter(
            subset["DATE"],
            subset["LAST_PRICE"],
            color=color,
            label=f"{sentiment} Day",
            alpha=0.6,
            s=60,
            edgecolors="black",
            linewidths=0.5
        )

    # Remove duplicate legend entries
    handles, labels = ax.get_legend_handles_labels()
    unique = dict(zip(labels, handles))
    ax.legend(unique.values(), unique.keys())

    # Summary stats
    avg_price = plot_df["LAST_PRICE"].mean()
    pos_pct = (plot_df["Dominant_Sentiment"] == "Positive").mean() * 100
    ax.set_title(f"{stock_name} Daily Sentiment & Price (Avg: {avg_price:.2f}, Pos Days: {pos_pct:.1f}%)")

    ax.set_xlabel("Date")
    ax.set_ylabel("Last Price")
    plt.tight_layout()
    plt.show()

# Generate charts for each top stock
for stock in top_10_stocks:
    plot_stock_with_sentiment_density(df, stock)
    plot_stock_sentiment(df, stock)
```

```python
# Now I will have a look at the data and how it supports my hypothesis

# Filter dataset to only include tweets from 2018 onwards
df_filtered = df[df["DATE"] >= pd.Timestamp("2018-01-01")].copy()

# AI Assisted me with making this look nice and neat

# Aggregate daily sentiment and stock price per company
# - Average LSTM_POLARITY gives a continuous sentiment score
# - Mode of LSTM_SENTIMENT gives the dominant sentiment label (Positive/Negative)
# - Average LAST_PRICE gives the daily closing price
daily_df = df_filtered.groupby(["STOCK", "DATE"]).agg({
    "LSTM_POLARITY": "mean",
    "LSTM_SENTIMENT": lambda x: x.mode()[0] if not x.mode().empty else None,
    "LAST_PRICE": "mean"
}).reset_index()

# Calculate next-day price change and direction
# - Helps assess whether sentiment may have influenced price movement
daily_df.sort_values(["STOCK", "DATE"], inplace=True)
daily_df["NEXT_PRICE"] = daily_df.groupby("STOCK")["LAST_PRICE"].shift(-1)
daily_df["PRICE_CHANGE"] = daily_df["NEXT_PRICE"] - daily_df["LAST_PRICE"]
daily_df["PRICE_DIRECTION"] = daily_df["PRICE_CHANGE"].apply(
    lambda x: "Up" if x > 0 else ("Down" if x < 0 else "Flat")
)

# Define logic to compare sentiment with price movement
# - Returns 'Aligned' if sentiment and price move in same direction
# - Returns 'Opposed' if they move in opposite directions
# - Returns 'Neutral' if price doesn't change
def sentiment_effect(row):
    if row["LSTM_SENTIMENT"] == "Positive" and row["PRICE_DIRECTION"] == "Up":
        return "Aligned"
    elif row["LSTM_SENTIMENT"] == "Negative" and row["PRICE_DIRECTION"] == "Down":
        return "Aligned"
    elif row["PRICE_DIRECTION"] == "Flat":
        return "Neutral"
    else:
        return "Opposed"

# End of AI assistance

# Apply sentiment effect logic to each row
daily_df["SENTIMENT_EFFECT"] = daily_df.apply(sentiment_effect, axis=1)

# Select relevant columns for analysis and reporting
result_df = daily_df[[
    "STOCK", "DATE", "LSTM_SENTIMENT", "LSTM_POLARITY",
    "LAST_PRICE", "NEXT_PRICE", "PRICE_CHANGE",
    "PRICE_DIRECTION", "SENTIMENT_EFFECT"
]]

# Tabulate how often sentiment aligns, opposes, or has no effect on price
summary = result_df.groupby("SENTIMENT_EFFECT").size().reset_index(name="Count")
print(summary)

# Visualize sentiment effect distribution as a bar chart
sns.countplot(data=result_df, x="SENTIMENT_EFFECT", palette="Set2")
plt.title("Tweet Sentiment vs. Stock Price Movement")
plt.xlabel("Sentiment Effect")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
